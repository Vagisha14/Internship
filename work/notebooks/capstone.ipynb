Research Question:

Which content pages are most worth reviewing or refreshing based on their search-performance signals, and can a scoring model rank these pages into actionable review priorities?

Decision Supported:

The analysis supports content teams in prioritizing which pages should be reviewed, refreshed, improved, or monitored first.

## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

### Data

This capstone uses the public-safe FlyRank Internship Starter dataset, `FlyRank/internship-starter`, which contains 30,000 anonymized content-performance rows across 32 pseudonymized clients and 53 fields.

The analysis uses content-level search and engagement signals measured over a 90-day window, including impressions, clicks, sessions, engagement, content age, freshness, CTR, average position, and recent versus previous 30-day performance.

For the refresh/opportunity analysis, the data is treated as anonymized content-performance data. Client identifiers, content identifiers, URLs, titles, keywords, domains, and other potentially identifying information are not used in public reporting.

Leakage-prone product flags and precomputed recommendation fields are excluded from model features, including health scores and action/opportunity flags, because they can encode the outcome being predicted.

The analysis focuses on public-safe aggregate patterns and ranked recommendations rather than exposing individual client or content details.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

The analysis uses a content refresh and opportunity scoring approach to identify pages that may be worth reviewing or refreshing. First, the available content-performance and search-related signals are cleaned and checked for missing or inconsistent values. Relevant features such as impressions, clicks, sessions, engagement, content age, freshness, CTR, average position, and recent versus previous performance are used to understand content performance. Leakage-prone fields and precomputed recommendation or action fields are excluded from the analysis. A scoring model is then used to combine these performance signals into a ranked priority score. The pages are ranked according to this score so that content teams can focus their review and refresh efforts on the highest-priority opportunities. The results are interpreted as decision-support recommendations rather than guaranteed outcomes.

## 5. Limitations

*What this work cannot claim.*

This analysis is based on anonymized search and content-performance data and therefore may not capture all factors that influence content performance. The scoring approach identifies patterns and prioritization opportunities, but it does not establish causal relationships between content changes and future performance. Model results may also be affected by data quality, missing values, limited historical information, and the selected features. The recommendations should therefore be treated as decision-support signals rather than guaranteed outcomes. Further validation on new data and testing over time would be required before using the scoring system as a production decision-making tool.
