```python
import matplotlib.pyplot as plt


time = [6, 8, 10, 12, 14, 16, 18, 20, 22]
temperature = [22, 24, 27, 31, 34, 33, 30, 27, 25]


max_temp = max(temperature)
min_temp = min(temperature)

max_time = time[temperature.index(max_temp)]
min_time = time[temperature.index(min_temp)]


plt.figure(figsize=(8, 5))

plt.plot(time, temperature, marker='o', linewidth=2)
plt.annotate(
    f'Max: {max_temp}°C',
    xy=(max_time, max_temp),
    xytext=(max_time, max_temp + 2),

)


plt.annotate(
    f'Min: {min_temp}°C',
    xy=(min_time, min_temp),
    xytext=(min_time, min_temp - 3),

)
plt.annotate(
    f"Maximum = {max_temp}°C",
    xy=(max_time, max_temp),
    xytext=(10, 10),
    textcoords="offset points"
)

plt.xlabel('Time (hours)')
plt.ylabel('Temperature (°C)')
plt.title('Temperature Variation with Time')

plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

kolkata = [20, 23, 27, 31, 33, 31, 30, 30, 30, 29, 25, 21]
delhi = [14, 18, 24, 30, 34, 33, 32, 31, 30, 27, 21, 16]
mumbai = [25, 26, 28, 30, 31, 29, 28, 28, 29, 29, 27, 26]

plt.figure(figsize=(12, 12))

plt.plot(months, kolkata, marker='o', label='Kolkata')
plt.plot(months, delhi, marker='o', label='Delhi')
plt.plot(months, mumbai, marker='o', label='Mumbai')

all_data = {
    "Kolkata": kolkata,
    "Delhi": delhi,
    "Mumbai": mumbai
}

max_city = max(all_data, key=lambda city: max(all_data[city]))
max_temp = max(all_data[max_city])
max_month = months[all_data[max_city].index(max_temp)]

min_city = min(all_data, key=lambda city: min(all_data[city]))
min_temp = min(all_data[min_city])
min_month = months[all_data[min_city].index(min_temp)]

plt.annotate(
    f"Maximum: {max_temp}°C ({max_city}, {max_month})",
    xy=(max_month, max_temp),
    xytext=(20, 20),
    textcoords="offset points",

)

plt.annotate(
    f"Minimum: {min_temp}°C ({min_city}, {min_month})",
    xy=(min_month, min_temp),
    xytext=(20, -30),
    textcoords="offset points",

)

plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.title("Monthly Temperature Comparison: Kolkata, Delhi and Mumbai")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```
